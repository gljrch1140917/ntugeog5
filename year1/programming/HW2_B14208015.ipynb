### Question 1

```python
import numpy as matrix

grid = matrix.zeros((15, 15), int) 
for r in range(15): 
    for c in range(15):
        dist = min(r, c, 14-r, 14-c)
        if dist < 2:    
            grid[r, c] = 1
        elif dist < 5:  
            grid[r, c] = 2
        else:          
            grid[r, c] = 3

noise_idx = matrix.random.choice(15*15, 15, replace=False)
rows = noise_idx // 15  
cols = noise_idx % 15   
grid[rows, cols] = 0  

for r in range(15):
    for c in range(15):
        if grid[r, c] == 0:
            dist = min(r, c, 14-r, 14-c)
            if dist < 2: grid[r, c] = 1
            elif dist < 5: grid[r, c] = 2
            else: grid[r, c] = 3

vals, counts = matrix.unique(grid, return_counts=True)

pixel_stats = {f"Layer {int(v)}": int(c) for v, c in zip(vals, counts) if v != 0}

print(f"Result: {pixel_stats}")
```

### Question 2

```python
def collect_water(grid, r, c, visited_set): 
    visited_set.add((r, c)) 
    current_elevation = grid[r][c] 
    
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = r + dr, c + dc 
        if 0 <= nr < len(grid) and 0 <= nc < len(grid[0]):
            if (nr, nc) not in visited_set and grid[nr][nc] <= current_elevation:
                collect_water(grid, nr, nc, visited_set)

terrain = [[50, 50, 50, 30, 30], [50, 40, 40, 30, 20], [50, 40, 10, 10, 20], [50, 40, 40, 30, 20], [50, 50, 50, 50, 50]]

flooded_coords = set()
collect_water(terrain, 2, 2, flooded_coords)
total_area = len(flooded_coords)
avg_elev = sum(terrain[row][col] for row, col in flooded_coords) / total_area

print(f"flooded area size: {total_area}, average elevation: {avg_elev:.2f}")
```

### Question 3

```python
import numpy as matrix

station_dict = {
    "Taipei": (25.1, 121.5),
    "Taichung": (24.1, 120.6),
    "Kaohsiung": (22.6, 120.3),
    "Hualien": (23.9, 121.6),
    "Tainan": (23.0, 120.2)}

names = list(station_dict.keys())
coords = matrix.array(list(station_dict.values()))
my_location = matrix.array([24.5, 121.0])

diff = coords - my_location 
dist = matrix.sqrt(matrix.sum(diff**2, axis=1))

mask = dist < 1.2
nearby_stations = [names[i] for i, is_near in enumerate(mask) if is_near]

print(f"Stations within 1.2 units: {nearby_stations}")
```
