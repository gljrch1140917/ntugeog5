### Question 1

```python
correct_pin = 1910
attempts = 0

while attempts < 3:
    try:
        password = int(input("Please enter your PIN:"))
        if password == correct_pin:
            print("Access Granted")
            break
        else:
            attempts += 1
            print(f"Incorrect PIN. Attempts remaining: {3 - attempts}")
            if attempts == 3:
                print("Account Locked")
    except:
        print("An invalid format try again")


if password == correct_pin:
    balance = 10000
    is_running = True
    
    while is_running:
        try:
            withdraw_amount = int(input("Please enter the amount to withdraw: "))
            if withdraw_amount % 500 == 0:
                if withdraw_amount <= 4000 and withdraw_amount <= balance:
                    balance -= withdraw_amount
                    print(f"Transaction successful! Remaining balance: {balance}")
                    
                    while True:
                        again = input("Do you want to perform another transaction? (Y/N): ").upper()
                        if again == "Y":
                            break
                        elif again == "N":
                            print("Thank you, please take your card.")
                            is_running = False 
                            break           
                        else:
                            print("Invalid input! Please enter only 'Y' or 'N'.")
                else:
                    print("Error: Amount must be <= 4000 and within your balance.")
            else:
                print("Error: Amount must be a multiple of 500.")
        except ValueError:
            print("Invalid input, please enter a number.")
```

### Question 2

```python
while True:
    try:
        N = int(input("Please enter a positive integer:"))
        if N > 1:
            break
    except:
        print("The number must be an integer that greater than 1.")

temp_n = N
divisor = 2

while True:
    count = 0
    if temp_n > 1:
        while temp_n % divisor == 0:
            temp_n = temp_n / divisor
            count += 1
                
        if count == 1:
            print(f"Factor {divisor} appears {count} time")
        else:
            if count > 1:
                print(f"Factor {divisor} appears {count} times")
                    
        divisor += 1
```

### Question 3

```python
def power(base, exp):
    if exp == 0:
        return 1
    
    if exp % 2 == 0:
        half_power = power(base, exp // 2)
        return half_power * half_power
    else:
        return base * power(base, exp - 1)

try:
    u_base = float(input("Enter the base: "))
    u_exp = int(input("Enter the exponent: "))

    if u_exp < 0:
        print("Error: Exponent must be a non-negative integer.")
    else:
        result = power(u_base, u_exp)
        print(f"Result: {result}")

except:
    print("Invalid input! Please enter a valid number.")
```
